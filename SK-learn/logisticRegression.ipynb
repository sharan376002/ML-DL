# logistic Regression


logistics regression model is used of thr binary classfication

the loss function used for the logisctics regression is binary cross entroy

l(y,y.hat) = -ylogy.hat + (1-y)log(1-y.hat)

sum of the loss function is the cost function

- it is the classfication mode
- supervised learning model

it uses the sigmoid function 
1/1+e^-Z          
where Z= Wx + B


for updating weights ,

w = w- L*dw
b = b - L* db

where

dw = 1/m  * (y.hat - Y).X

db = 1/m * (y.hat - Y)


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

```

```python
class LogisticRegression:
    def __init__(self,learning_rate,n_iteration):
        self.learning_rate = learning_rate
        self.n_iteration = n_iteration

        


    def fit(self, X, y):
        

        # no of features(n) and no of samples(m)
        self.m, self.n  = X.shape


        #intilise the weight /bias 

        self.weight = np.zeros(self.n)
        self.bias = 0

        self.x = X
        self.y = y

        for i in range(self.n_iteration):
            self.update_weights(X)




    def update_weights(self, X):
        Y_predicted = 1/ (1 + np.exp(-(np.dot(X, self.weight) + self.bias)))
        # calculate the gradient

        dw = (1 / self.m) * np.dot(X.T, (Y_predicted - self.y))
        db = (1/self.m) * (Y_predicted - self.y)

        self.weight -= self.learning_rate * dw
        self.bias -= self.learning_rate * db



    def predict(self, X):  

        y_hat = 1 / (1 + np.exp(-(np.dot(X, self.weight) + self.bias)))
        y_predicted = np.where(y_hat > 0.5, 1, 0)
        return y_predicted
    



```

## workflow for logistics regression

- set learning rate and no of iterations and random weights and bias value

- build logistic regression function sigmoid

- update the parameter using gradient descent

- finally we will get the best weight adn bias value it has minimus cost function 

- build the predict fumcttion to determine class of data point
