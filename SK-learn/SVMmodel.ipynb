```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# building the  SVM model

class SVM_classfier():
    # initializing the hyperparameters

    def __init__(self, learning_rate, n_iters, lambda_param):

        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.lambda_param = lambda_param

    # fit method to train the model
    def fit(self,X, y):
        # no of features(n) and no of samples(m)
        self.m, self.n  = X.shape


        #intilise the weight /bias 

        self.weight = np.zeros(self.n)
        self.bias = 0

        self.x = X
        self.y = y

        for i in range(self.n_iters):
            self.update_weights()


    # update the weights and bias values
    def update_weights(self):

        # label encoding
        y_label = np.where(self.y <=0 , -1,1)


        for index, X_i in enumerate(self.x):

            condition = y_label[index] * (np.dot(X_i, self.weight) - self.bias) >= 1

            if condition:

                dw = 2 * self.lambda_param * self.weight
                db = 0
            
            else:
                dw = 2 * self.lambda_param * self.weight - np.dot(X_i,y_label[index])


                db  = y_label[index]

            self.weight -= self.learning_rate * dw

            self.bias -= self.learning_rate * db
        


        #predicting the labels 0 or 1
    def predict(self):
        
        op = np.dot(X,self.weight) - self.bias

        predictions = np.sign(op)

        return np.where(predictions <= 0, 0, 1)
    




         
```

```python
model = SVM_classfier(learning_rate=0.01, n_iters=1000, lambda_param=0.01)
# X, y = load_data()  # Assuming you have a function to load your data
```
