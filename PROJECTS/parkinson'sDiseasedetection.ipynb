```python
x = pk.drop(columns = ['name','status'], axis=1)
y = pk['status']
```

```python
print(x)
```

```python
print(y)
```

spliting the data for the trainning and testing

```python
train_x , test_x,train_y,test_y = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(train_x)
```

```python
print(test_x)
```

```python
print(x.shape,train_x.shape,test_x.shape)
```

```python
# data standartization

std =  StandardScaler()
```

```python
std.fit(train_x)
```

```python
train_x = std.transform(train_x)

test_x = std.transform(test_x)
```

```python
#importing the svm model

model = svm.SVC(kernel='linear')
```

```python
model.fit(train_x,train_y)
```

```python
# to find the  accuracy of the trainning data

train_pred = model.predict(train_x)
acc= accuracy_score(train_pred,train_y)
print(acc)
```

```python
# accuracy score on training data
X_test_prediction = model.predict(test_x)
test_data_accuracy = accuracy_score(test_y, X_test_prediction)
print('Accuracy score of test data : ', test_data_accuracy)
```
